```python
from pathlib import Path

import pandas as pd

# Getting the csv rows I need, and inspecting
ROOT = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent
RAW_PATH = ROOT / "data" / "raw" / "adult24.csv"
OUT_PATH = ROOT / "data" / "processed" / "kristines_processed.csv"

df = pd.read_csv(RAW_PATH, usecols=["SLPHOURS_A", "SLPREST_A", "SLPFLL_A", "SLPSTY_A", "SEX_A", "PA18_02R_A", "SCHCURENR_A", "AGEP_A"])
df = df[(df["SCHCURENR_A"] == 1) & (df["AGEP_A"].between(18, 29))]
print(df.info())
print(df.isna().sum())
print(df.head())
print(df.shape) 
#print(df.duplicated().sum()) #Checking for duplicates is meaningless, as it's likely And allowed for a person to answer the same as another
```

```python
print(df["SLPHOURS_A"].value_counts()) #On average, how many hours of sleepn in 24 hours?

df["SLPHOURS_A"].replace({97: float("nan"), 98: float("nan"), 99: float("nan")})
df.rename(columns={"SLPHOURS_A": "nightly_sleep"}, inplace=True)
print(df["nightly_sleep"].value_counts())
```

```python
print(df["SLPREST_A"].value_counts()) #During last 30 days, how often woke up feeling rested?

df["SLPREST_A"] = df["SLPREST_A"].replace({7: float("nan"), 8: float("nan"), 9: float("nan")})

df["SLPREST_A"].map({1: "Never", 2: "Some Days", 3: "Most days", 4: "Every Day"})

df.rename(columns={"SLPREST_A": "monthly_fullyRested"}, inplace=True)
print(df["monthly_fullyRested"].value_counts())
```

```python
print(df["SLPFLL_A"].value_counts()) #During last 30 days, how often trouble falling asleep?

df["SLPFLL_A"] = df["SLPFLL_A"].replace({7: float("nan"), 8: float("nan"), 9: float("nan")})

df["SLPFLL_A"].map({1: "Never", 2: "Some Days", 3: "Most days", 4: "Every Day"})
df.rename(columns={"SLPFLL_A": "monthly_troubleSleeping"}, inplace=True)
print(df["monthly_troubleSleeping"].value_counts())
```

```python
print(df["SLPSTY_A"].value_counts()) #How often did you have trouble staying asleep?

df["SLPSTY_A"] = df["SLPSTY_A"].replace({7: float("nan"), 8: float("nan"), 9: float("nan")})

df["SLPSTY_A"].map({1: "Never", 2: "Some Days", 3: "Most days", 4: "Every Day"})
df.rename(columns={"SLPSTY_A": "monthly_troubleStayingAsleep"}, inplace=True)
print(df["monthly_troubleStayingAsleep"].value_counts())
```

```python
print(df["SEX_A"].value_counts()) #Are you male or female?

df["SEX_A"] = df["SEX_A"].replace({7: float("nan"), 9: float("nan")})

df["SEX_A"].map({1: "Male", 2: "Female",})
df.rename(columns={"SEX_A": "gender"}, inplace=True)
print(df["gender"].value_counts())
```

```python
print(df["PA18_02R_A"].value_counts()) #Physical activity met guidelines for aerobic activity

df["PA18_02R_A"] = df["PA18_02R_A"].replace({8: float("nan")})

df["PA18_02R_A"].map({1: "Inactive", 2: "Insufficiently active", 3: "Sufficiently active"})
df.rename(columns={"PA18_02R_A": "activityLevelPerGuidelines"}, inplace=True)
print(df["activityLevelPerGuidelines"].value_counts())
```

```python
df = df.drop(columns=["SCHCURENR_A", "AGEP_A"])
df.to_csv(OUT_PATH, index=False)
```
